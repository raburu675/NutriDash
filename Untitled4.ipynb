```python
from django.shortcuts import render
from django.http import HttpResponse
import plotly.express as px
import pandas as pd
from plotly.offline import plot 

data = {
    'days' : ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
    'calories' : [2100,2300,1950,1700,1900,2250,2134]
}
df = pd.DataFrame(data)

#define the target calories and add it to the dataframe
target_calories = 2170
df['target_calories'] = target_calories

df_melted = pd.melt(df, id_vars=['days'], value_vars=['calories', 'target_calories'],var_name='line_type', value_name='value')

#create the plotly chart using px.line
fig = px.line(df_melted, 
                  x='days', 
                  y='value', 
                  color='line_type', # Differentiate lines by color
                  line_shape='spline', # Create a smooth line for the trend
                  title='Weekly Calorie Intake vs. Target',
                  color_discrete_map={
                      'calories': '#4285F4',
                      'target_calories': '#DB4437'
                  })
 # 5. Customize the layout
fig.update_layout(
    title={
        'text': 'Weekly Calorie Intake vs. Target',
        'y':0.9,
        'x':0.5,
        'xanchor': 'center',
        'yanchor': 'top'
    },
    xaxis_title='Day of the Week',
    yaxis_title='Calories (kcal)',
    font=dict(
        family="Inter, sans-serif",
        size=12,
        color="#333"
    ),
    plot_bgcolor='rgba(0,0,0,0)',
    paper_bgcolor='rgba(0,0,0,0)',
    margin=dict(l=50, r=50, t=80, b=50),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1,
        title=None
    )
)

fig.show()
```

```python
import plotly.express as px
import pandas as pd
from plotly.offline import plot 

data = {
    'Months' : ['January','Feburary','March','April','May','June','JUly','August','September','October','November','December'],
    'weight_kg' : [74.5, 72.1, 65.3, 68, 69.8, 72.1, 70.5, 67.1, 71.3, 73.7, 73.8, 74.1]
}

weight_df = pd.DataFrame(data)
#create the area chart using px.area
fig = px.area(
    weight_df,
    x='Months',
    y='weight_kg',
    title='weight tracker',
    labels={'weight_kg': 'Weight (kg)', 'day': 'Day of the Week'},
    line_shape='spline'
)
fig.update_yaxes(range=[55, 76])
fig.show()
```
